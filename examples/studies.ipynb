# Combining Indicators into Studies (pandas only)



```python
import pandas as pd

from mintalib.samples import sample_prices
from mintalib.indicators import SMA, EMA, MACD
from mintalib.studies import Study
```

```python
prices = sample_prices()
prices
```

```python
study = Study(MACD(), sma50=SMA(50), sma200=SMA(200))

print(study)
```

```python

prices.pipe(study)
```

```python
prices.pipe(study, merge=True)
```

## Multi Symbol Dataset

```python

dataset = pd.concat({k:prices for k in ("AAA", "BBB", "CCC")}, names=["symbol"])    
dataset

```

```python
dataset.groupby("symbol", group_keys=False).apply(study)
```

```python
dataset.groupby("symbol", group_keys=False).apply(study, merge=True)
```
